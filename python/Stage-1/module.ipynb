```python
from math import sqrt,pi,pow,ceil,floor
a=int(input())
b=int(input())
```

```python
print(math.sqrt(a),math.sqrt(b))
print(math.pi)
print(math.pow(a,b))
print(math.floor(5.6))
print(math.ceil(5.6))
print(math.factorial(a))
print(math.gcd(a,b))
```
