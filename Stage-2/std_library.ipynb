```python
students = [
    ("Alice", "CS"),
    ("Bob", "IT"),
    ("Charlie", "CS"),
    ("David", "Mechanical"),
    ("Eve", "IT"),
    ("Frank", "CS")
]
```

```python
d=defaultdict(list)
for name,department in students:
    d[department].append(name)

for department, name in d.items():
    print(department,"->",name)
```

```python
queue = deque(["Alice", "Bob", "Charlie"])
queue.append("David")

queue.popleft()

queue.appendleft("Eve")

queue.pop()

print(queue)
```

```python
developers = ["Alice", "Bob", "Charlie", "David", "Eve"]
teams = combinations(developers, 2)
for team in teams:
    print(team)
```

```python
characters = ["A", "B", "C"]

arrangements = permutations(characters, 2)

for arrangement in arrangements:
    print(arrangement)
```

```python
from functools import *
```

```python
def multiply(a, b):
    return a * b

multiply_by_10 = partial(multiply, b=10)

print(multiply_by_10(5))
print(multiply_by_10(8))
```

```python
numbers = [2, 3, 4, 5]

result = reduce(lambda x, y: x * y, numbers)

print(result)
```

```python
@lru_cache
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(10))
```
